# Notebook 7: LangGraph Agent Evaluation

Build and test the churn copilot agent on sample business questions.

```python
import sys
from pathlib import Path

ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT))

from genai.copilot import ask, build_graph
```

```python
graph = build_graph()
print('LangGraph agent compiled:', graph is not None)
```

```python
TEST_QUESTIONS = [
    'Why are customers churning?',
    'Which segment has highest churn?',
    'What is the churn trend last 3 months?',
    'Who are the top 10 customers at risk?',
    'What is ARPU by plan?',
    'How does churn differ for senior vs non-senior?',
    'Which city has highest churn?',
    'What is the overall churn rate?',
]
```

```python
results = []
for q in TEST_QUESTIONS:
    print('=' * 60)
    print('Q:', q)
    out = ask(q)
    answer = out['answer']
    print('A:', answer[:500], '...' if len(answer) > 500 else '')
    has_number = any(c.isdigit() for c in answer)
    results.append({'question': q, 'ok': bool(answer) and has_number, 'has_chart': out.get('chart') is not None})

import pandas as pd
pd.DataFrame(results)
```
