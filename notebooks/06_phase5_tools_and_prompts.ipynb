```python
sql = """
SELECT churn_reason, COUNT(*) AS cnt
FROM telco_customers
WHERE churn_label = 'Yes'
GROUP BY churn_reason
ORDER BY cnt DESC
LIMIT 10
"""
print(run_sql_query.invoke({'query': sql}))
```

```python
chart_sql = """
SELECT plan_type, ROUND(AVG(churn_value)*100,2) AS churn_rate_pct
FROM vw_powerbi_customers
GROUP BY plan_type
ORDER BY churn_rate_pct DESC
"""
print(generate_chart.invoke({
    'chart_type': 'bar',
    'sql_query': chart_sql,
    'title': 'Churn Rate by Plan'
}))
```
